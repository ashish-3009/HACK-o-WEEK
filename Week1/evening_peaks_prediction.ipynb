```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import plotly.graph_objects as go
```

```python
df = pd.read_csv("/content/electricityConsumptionAndProductioction.csv")
```

```python
df['DateTime'] = pd.to_datetime(df['DateTime'], format="%Y-%m-%d %H:%M:%S")
df = df.sort_values('DateTime')
df.set_index('DateTime', inplace=True)
```

```python
df['hour'] = df.index.hour
df['weekday'] = df.index.dayofweek
df['week'] = df.index.isocalendar().week
df['year'] = df.index.year
```

```python
df['ema_24h'] = df['Consumption'].ewm(span=24).mean()
```

```python
evening_df = df.between_time("17:00", "21:00")
```

```python
weekly_peaks = (
    evening_df
    .resample('W')['Consumption']
    .max()
    .to_frame(name='evening_peak')
)
```

```python
weekly_peaks['week'] = weekly_peaks.index.isocalendar().week
weekly_peaks['year'] = weekly_peaks.index.year
```

```python
weekly_features = (
    df.resample('W')[['Nuclear', 'Wind', 'Hydroelectric', 'Oil and Gas', 'Coal', 'Solar', 'Biomass']]
    .mean()
)
```

```python
weekly_features['avg_consumption'] = df['Consumption'].resample('W').mean()
```

```python
weekly_df = weekly_peaks.merge(weekly_features, left_index=True, right_index=True)
```

```python
X = weekly_df[['avg_consumption', 'Nuclear', 'Wind', 'Hydroelectric', 'Oil and Gas', 'Coal', 'Solar', 'Biomass']]
y = weekly_df['evening_peak']
```

```python
model = LinearRegression()
model.fit(X, y)
weekly_df['predicted_peak'] = model.predict(X)
```

```python
fig = go.Figure()
```

```python
fig.add_trace(go.Scatter(
    x=df.index,
    y=df['Consumption'],
    name="Raw Consumption",
    opacity=0.3
))
```

```python
fig.add_trace(go.Scatter(
    x=df.index,
    y=df['ema_24h'],
    name="Smoothed (EMA 24h)",
    line=dict(width=3)
))
```

```python
fig.update_layout(
    title="Hourly Electricity Consumption (Smoothed)",
    xaxis_title="Time",
    yaxis_title="MW"
)
```

```python
fig.show()
```

```python
fig2 = go.Figure()
```

```python
fig2.add_trace(go.Scatter(
    x=weekly_df.index,
    y=weekly_df['evening_peak'],
    mode='markers',
    name='Actual Evening Peak'
))
```

```python
fig2.add_trace(go.Scatter(
    x=weekly_df.index,
    y=weekly_df['predicted_peak'],
    mode='lines',
    name='Predicted Peak'
))
```

```python
fig2.update_layout(
    title="Weekly Evening Peak Prediction",
    xaxis_title="Week",
    yaxis_title="MW"
)
```

```python
fig2.show()
```

```python
y_true = weekly_df['evening_peak']
y_pred = weekly_df['predicted_peak']
```

```python
mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_true, y_pred)
```

```python
print(f"MAE: {mae:.2f} MW")
print(f"RMSE: {rmse:.2f} MW")
print(f"R²: {r2:.3f}")
```
